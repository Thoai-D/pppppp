```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

model_path = r"C:\Users\thoai\.cache\kagglehub\models\krishnagarigipati\pneumonia-detector\keras\v1\1\final_pneumonia_model.h5"
test_dir = r"C:\SUTD\50.021 Artificial Intelligence\pppppp\New folder\chest_xray\test"
# test_dir = r"C:\SUTD\50.021 Artificial Intelligence\pppppp\Master_Dataset\test

model = tf.keras.models.load_model(model_path, compile=False)
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

test_datagen = ImageDataGenerator(rescale=1./255)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(224, 224),
    color_mode="rgb",
    batch_size=32,
    class_mode="binary",
    shuffle=False
)

loss, acc = model.evaluate(test_generator)

print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {acc:.4f}")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score

pred_probs = model.predict(test_generator)
preds = (pred_probs > 0.5).astype(int).reshape(-1)
y_true = test_generator.classes

acc_normal = accuracy_score(y_true, preds)
acc_flipped = accuracy_score(y_true, 1 - preds)

print("Normal accuracy :", acc_normal)
print("Flipped accuracy:", acc_flipped)
```
